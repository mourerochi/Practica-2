1. Diagnostico inicial

```python
import pandas as pd
import numpy as np

df = pd.read_csv("practica_2.csv")
print("Cantidad original:", df.shape)
print("Tipo de datos:", df.dtypes)
```

2. Ajustes

```python
print("Cantidad de duplicados:", df.duplicated().sum())
df = df.drop_duplicates()
print("Cantidad actualizada:", df.shape)

print(df.columns)
df["Datetime"] = pd.to_datetime(df["Datetime"])
print("Actualización de los tipos de datos:", df.dtypes)
```

3. Análisis de tiempo de registros

```python
print(df["Datetime"])
periodo_total = df["Datetime"].max() - df["Datetime"].min()
print("El periodo de tiempo total que hay es de:", periodo_total)
df["Datetime"].diff()
```

Usamos diff() para calcular la diferencia entre cada valor y el valor anterior, para poder saber si hay gaps, lo cual si hay.
