1. Diagnostico inicial

```python
import pandas as pd
import numpy as np

df = pd.read_csv("https://raw.githubusercontent.com/mourerochi/Practica-2/refs/heads/main/practica_2.csv")
print("Cantidad original:", df.shape)
print("Tipo de datos:", df.dtypes)
```

2. Ajustes

```python
print("Cantidad de duplicados:", df.duplicated().sum())
df = df.drop_duplicates()
print("Cantidad actualizada:", df.shape)

print(df.columns)
df["Datetime"] = pd.to_datetime(df["Datetime"])
print("Actualización de los tipos de datos:", df.dtypes)
```

3. Análisis de tiempo de registros

```python
print(df["Datetime"])
periodo_total = df["Datetime"].max() - df["Datetime"].min()
print("El periodo de tiempo total que hay es de:", periodo_total)
df["Datetime"].diff()
```

Usamos diff() para calcular la diferencia entre cada valor y el valor anterior, para poder saber si hay gaps, lo cual si hay.

4. Variables Sensadas

Las variables que sean han medido son Temperatura, Humedad, Presión, CO2, PM2.5 y PM10. El significado fisico y las unidades respectivas de cada una de las variables es, la temperatura mide el grado de calor del ambiente y se expresa en °C. La humedad mide la humedad relativa del aire y se expresa en porcentaje (%). La presión mide presión atmosférica y se expresa en hPa. El CO2 mide la concentración de dióxido de carbono y se expresa en ppm. PM2.5 y PM10 miden la concentración de partículas suspendidas en el aire y se expresan en µg/m3.
